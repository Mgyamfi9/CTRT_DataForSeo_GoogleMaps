```python
import numpy as np
conditions = [
    df["Rating"] >= 4.0,
    (df["Rating"] >= 3.0) & (df["Rating"] < 4.0)
]

choices = [
    "High",
    "Medium"
]

# Apply conditions using numpy select
df["Rate"] = np.select(conditions, choices, default="Low")
```
